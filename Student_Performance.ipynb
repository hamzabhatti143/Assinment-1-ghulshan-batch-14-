```python
import pandas as pd
```

```python
data = pd.read_csv("Student_Performance.csv")

data.head(5)
```

```python
data.info()
```

```python
data.plot(kind="scatter",x=['Previous Scores'],y=['Performance Index'],
         title="Student Performance")
```

```python
data.isna().sum()
```

```python
x = data[['Previous Scores']]
y = data[['Performance Index']]
```

```python
print(x.shape)
print(y.shape)
```

```python
print(x.head())
print(y.head())
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x,y, test_size=0.20, random_state=32)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
y_train.shape, y_test.shape
```

```python
from sklearn.linear_model import LinearRegression

performance_predictor = LinearRegression()
```

```python
performance_predictor.fit(x_train, y_train)
```

```python
performance_predictor.coef_
```

```python
performance_predictor.intercept_
```

```python
y_pred_train = performance_predictor.predict(x_train)
```

```python
import matplotlib.pyplot as plt
```

```python
data.plot(kind="scatter", x="Previous Scores", y="Performance Index", color="y")
plt.plot(x_train, y_pred_train, color="r")
```

```python
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
```

```python
import math
math.sqrt(mean_squared_error(y_train, y_pred_train))
```

```python
r2_score(y_train, y_pred_train)
```

```python
mean_absolute_error(y_train, y_pred_train)
```

```python
y_pred_test = performance_predictor.predict(x_test)
```

```python
import math
math.sqrt(mean_squared_error(y_test, y_pred_test))
```

```python
r2_score(y_test, y_pred_test)
```

```python
mean_absolute_error(y_test, y_pred_test)
```

```python
def performace_pred(previous_score):
    return performance_predictor.predict([[previous_score]])
```

```python
performace_pred(70)
```

```python
performance_predic = float(input("Enter Your Performace Score: "))

performace_pred(performance_predic)
```
