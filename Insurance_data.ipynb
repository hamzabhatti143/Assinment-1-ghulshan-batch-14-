```python
import pandas as pd
```

```python
data = pd.read_csv("insurance data.csv")

data.head(5)
```

```python
data.info()
```

```python
data.isna().sum()
```

```python
x = data[['age', 'sex', 'bmi', 'children', 'smoker', 'region']]
y = data['charges']
```

```python
x = pd.get_dummies(x, drop_first=True)
```

```python
x.head(5)
```

```python
y.head(5)
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x,y, test_size=0.20, random_state=32)
```

```python
print(x.head())
print(y.head())
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
y_train.shape, y_test.shape
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
```

```python
lr.fit(x_train, y_train)
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
y_pred_train = lr.predict(x_train)
```

```python
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
```

```python
import math
math.sqrt(mean_squared_error(y_train, y_pred_train))
```

```python
r2_score(y_train, y_pred_train)
```

```python
mean_absolute_error(y_train, y_pred_train)
```

```python
y_pred_test = lr.predict(x_test)
```

```python
import math
math.sqrt(mean_squared_error(y_test, y_pred_test))
```

```python
r2_score(y_test, y_pred_test)
```

```python
mean_absolute_error(y_test, y_pred_test)
```

```python
def lr_pred(age, sex, bmi, children, smoker, region):
    input_data = pd.DataFrame({
        'age': [age],
        'bmi': [bmi],
        'children': [children],
        'sex_male': [sex == 'male'],
        'smoker_yes': [smoker == 'yes'],
        'region_northwest': [region == 'northwest'],
        'region_southeast': [region == 'southeast'],
        'region_southwest': [region == 'southwest']
    })

    return lr.predict(input_data)
```

```python
lr_pred(30, 'male', 25.5, 1, 'no', 'southeast')
```

```python
age = int(input("Enter your age: "))
sex = input("Enter your sex (male/female): ").lower()
bmi = float(input("Enter your BMI: "))
children = int(input("Enter number of children: "))
smoker = input("Are you a smoker? (yes/no): ").lower()
region = input("Enter your region (southwest/southeast/northwest/northeast): ").lower()

prediction = lr_pred(age, sex, bmi, children, smoker, region)

print("\n========== Insurance Prediction ==========")
print(f"Age              : {age}")
print(f"Sex              : {sex}")
print(f"BMI              : {bmi}")
print(f"Children         : {children}")
print(f"Smoker           : {smoker}")
print(f"Region           : {region}")
print("------------------------------------------")
print(f"Predicted Charges : ${prediction}")
print("==========================================")
```
