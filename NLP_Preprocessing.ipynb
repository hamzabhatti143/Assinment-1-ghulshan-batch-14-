```python
import pandas as pd
import string
import re
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
data.reviews.fillna(data.reviews.mode()[0], inplace=True)
```

```python
def text_preprocessor(text):
  text = text.lower()
  text = re.sub("\s+", " ", text)
  text = " ".join([a for a in text.split() if a not in string.punctuation])
  text = " ".join([a for a in text.split() if a not in stopwords.words("english")])
  return text
```

```python
def token_divider(text):
  tok = Tokenizer()
  tok.fit_on_texts([text])
  new_text = tok.texts_to_sequences([text])
  new_text = tok.sequences_to_texts(new_text)
  print("sequence: ", new_text)
  print("tokens: ", tok.word_index)
  return new_text
```

```python
user_text = input("Write some text: ")

text = text_preprocessor(user_text)
token_divider(text)
```
